```python
# Step 2: API కీ లేని ఓపెన్ సోర్స్ AI ని రన్ చేయడం
from transformers import pipeline

print("గూగుల్ క్లౌడ్ ద్వారా AI మోడల్ లోడ్ అవుతోంది...")

# Hugging Face నుండి ఫ్రీ మోడల్‌ని లోడ్ చేస్తున్నాం
ai_brain = pipeline("text-generation", model="gpt2")

# AI కి మనం ఇచ్చే ప్రశ్న / ప్రాంప్ట్
my_question = "Artificial Intelligence can help students to"

# AI జవాబును జనరేట్ చేస్తుంది
response = ai_brain(my_question, max_length=50, num_return_sequences=1)

print("\n------------------------------------")
print("AI తయారు చేసిన ఆన్సర్:")
print(response[0]['generated_text'])
print("------------------------------------")
```

```python
# Step 3: AI మాటలను ఆడియో (వాయిస్) గా మార్చే కోడ్
!pip install gTTS

from gtts import gTTS
from IPython.display import Audio

# AI చెప్పాల్సిన సమాధానం
speech_text = "Artificial Intelligence can help students to make better decisions."

# టెక్స్ట్‌ను ఆడియోగా మార్చడం
tts = gTTS(text=speech_text, lang='en')
tts.save("ai_voice.mp3")

print("AI మాట్లాడబోతోంది, వినండి:")
# కొలాబ్‌లోనే ఆడియో ప్లే చేయడం
Audio("ai_voice.mp3", autoplay=True)
```

```python
# Step 4: Interactive Visual AI Avatar UI

from IPython.display import HTML, display
import base64

# Step 3లో తయారైన audio file
with open("ai_voice.mp3", "rb") as audio_file:
    audio_data = base64.b64encode(audio_file.read()).decode("utf-8")

audio_src = "data:audio/mpeg;base64," + audio_data

avatar_ui = """
<div style="
    text-align:center;
    font-family:Arial,sans-serif;
    background:linear-gradient(135deg,#1e1e2f,#11111d);
    padding:30px;
    border-radius:20px;
    color:white;
    width:350px;
    margin:auto;
    box-shadow:0 10px 25px rgba(0,0,0,0.5);
">

    <h2 style="color:#00d2ff;">
        🤖 Interactive AI Avatar
    </h2>

    <!-- Avatar -->
    <div id="avatar" style="
        width:140px;
        height:140px;
        background:#252a44;
        border-radius:50%;
        margin:20px auto;
        border:4px solid #00d2ff;
        display:flex;
        flex-direction:column;
        align-items:center;
        justify-content:center;
        box-shadow:0 0 20px #00d2ff;
    ">

        <!-- Eyes -->
        <div style="
            display:flex;
            gap:30px;
            margin-bottom:20px;
        ">
            <div style="
                width:18px;
                height:18px;
                background:#00d2ff;
                border-radius:50%;
            "></div>

            <div style="
                width:18px;
                height:18px;
                background:#00d2ff;
                border-radius:50%;
            "></div>
        </div>

        <!-- Mouth -->
        <div id="mouth" style="
            width:45px;
            height:8px;
            background:#ff4757;
            border-radius:10px;
            transition:all 0.15s;
        "></div>

    </div>

    <!-- Status -->
    <p id="status" style="
        color:#a4b0be;
        font-weight:bold;
    ">
        ▶️ Play నొక్కండి
    </p>

    <!-- Audio -->
    <audio id="audioPlayer" controls style="width:100%;">
        <source src="AUDIO_SOURCE_HERE" type="audio/mpeg">
    </audio>

</div>

<script>

const audio = document.getElementById("audioPlayer");
const mouth = document.getElementById("mouth");
const status = document.getElementById("status");

let talkInterval;

audio.addEventListener("play", function() {

    status.innerText = "🗣️ AI మాట్లాడుతోంది...";
    status.style.color = "#00d2ff";

    talkInterval = setInterval(function() {

        let height = Math.floor(Math.random() * 22) + 6;

        mouth.style.height = height + "px";
        mouth.style.borderRadius = "50%";

    }, 120);

});

audio.addEventListener("pause", function() {

    clearInterval(talkInterval);

    mouth.style.height = "8px";
    mouth.style.borderRadius = "10px";

    if (!audio.ended) {
        status.innerText = "⏸️ AI Pause అయింది";
        status.style.color = "#ffa502";
    }

});

audio.addEventListener("ended", function() {

    clearInterval(talkInterval);

    mouth.style.height = "8px";
    mouth.style.borderRadius = "10px";

    status.innerText = "😊 AI Ready";
    status.style.color = "#2ed573";

});

</script>
"""

# Audio fileని HTMLలోకి connect చేయడం
avatar_ui = avatar_ui.replace("AUDIO_SOURCE_HERE", audio_src)

display(HTML(avatar_ui))
```
